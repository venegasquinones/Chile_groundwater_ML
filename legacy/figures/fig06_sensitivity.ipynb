# Figure 6 — Sensitivity analysis for temporal forecasting

**RIFT framework — groundwater level forecasting (Maipo River Basin, Chile)**

Composite of the former Figures 17 and 18.

- **(A)** Test $R^2$, RMSE, and MAPE vs. the minimum number of records per well.
- **(B)** Test $R^2$, RMSE, and MAPE vs. the train–test ratio.

The ensembles (Extra Trees, Random Forest) are robust to moderate record
lengths, while the Decision Tree degrades; the Decision Tree error plateaus
beyond a 60/40 split.

---
### How to use
1. Run the **imports / style** cell.
2. In the **Data** cell, paste the exact values from your results
   (`results_min_records.csv` / `results_ratio.csv` or the in-memory
   DataFrames). The arrays below were digitized from the original figures so
   the notebook renders out-of-the-box — **replace them with your authoritative
   numbers before submission.**
3. Run the **Build figure** cell. It writes vector **PDF** (preferred by
   Elsevier), plus 600-dpi **PNG** and **TIFF** to `OUTDIR`.

### Elsevier compliance notes
- Double-column width = 190 mm = 7.48 in (used here). Single column = 90 mm.
- Color line art: ≥ 300 dpi; we export vector PDF (resolution-independent) and
  600-dpi raster copies.
- Sans-serif fonts (Arial/Helvetica), `pdf.fonttype = 42` embeds editable
  TrueType so the production team can edit text if needed.


````python
import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.ticker import FuncFormatter, NullFormatter
from pathlib import Path

# ----- Output directory (edit to your project folder if desired) -----
OUTDIR = Path(".")
OUTDIR.mkdir(parents=True, exist_ok=True)

# ----- Elsevier-oriented rcParams -----
mpl.rcParams.update({
    "figure.dpi": 150,
    "savefig.dpi": 600,
    "font.family": "sans-serif",
    "font.sans-serif": ["Arial", "Helvetica", "DejaVu Sans"],
    "font.size": 8,
    "axes.titlesize": 9,
    "axes.labelsize": 8,
    "xtick.labelsize": 7,
    "ytick.labelsize": 7,
    "legend.fontsize": 8,
    "axes.linewidth": 0.8,
    "lines.linewidth": 1.4,
    "lines.markersize": 4,
    "xtick.major.width": 0.8,
    "ytick.major.width": 0.8,
    "axes.grid": True,
    "grid.alpha": 0.35,
    "grid.linewidth": 0.6,
    "axes.axisbelow": True,
    "savefig.bbox": "tight",
    "savefig.pad_inches": 0.03,
    "pdf.fonttype": 42,   # embed TrueType -> editable text in the PDF
    "ps.fonttype": 42,
})

# ----- Model registry (consistent colour + marker per model) -----
MODELS  = ["Decision Tree", "Extra Trees", "Random Forest"]
COLORS  = {"Decision Tree": "#1f77b4", "Extra Trees": "#ff7f0e", "Random Forest": "#2ca02c"}
MARKERS = {"Decision Tree": "o",       "Extra Trees": "s",       "Random Forest": "^"}
````

````python
# =====================================================================
# PANEL A — minimum records per well  (x is log-scaled)
# =====================================================================
min_records = np.array([3, 9, 27, 54, 81])

r2_A = {
    "Decision Tree": [0.868, 0.862, 0.888, 0.657, 0.352],
    "Extra Trees":   [0.924, 0.925, 0.927, 0.930, 0.930],
    "Random Forest": [0.920, 0.915, 0.926, 0.909, 0.840],
}
rmse_A = {
    "Decision Tree": [6.60, 6.68, 6.05, 10.40, 13.80],
    "Extra Trees":   [4.70, 4.60, 4.66, 4.63, 4.48],
    "Random Forest": [5.10, 5.22, 5.00, 5.30, 6.90],
}
mape_A = {
    "Decision Tree": [32.2, 34.0, 35.2, 53.8, 47.2],
    "Extra Trees":   [40.4, 48.6, 40.9, 41.7, 52.4],
    "Random Forest": [29.4, 30.2, 27.5, 29.6, 36.2],
}

# =====================================================================
# PANEL B — train / test ratio  (categorical, ordered 10/90 -> 90/10)
# =====================================================================
ratio_labels = ["10/90", "20/80", "30/70", "40/60", "50/50",
                "60/40", "70/30", "80/20", "90/10"]
ratio_x = np.arange(len(ratio_labels))

r2_B = {
    "Decision Tree": [-0.52, 0.74, 0.75, 0.77, 0.85, 0.88, 0.89, 0.89, 0.92],
    "Extra Trees":   [ 0.79, 0.87, 0.88, 0.89, 0.92, 0.93, 0.94, 0.94, 0.95],
    "Random Forest": [ 0.56, 0.86, 0.88, 0.89, 0.92, 0.93, 0.94, 0.94, 0.96],
}
rmse_B = {
    "Decision Tree": [20.3, 8.5, 8.4, 8.1, 6.6, 6.0, 5.8, 6.0, 5.0],
    "Extra Trees":   [ 7.5, 5.8, 5.7, 5.3, 4.8, 4.4, 4.3, 4.2, 3.8],
    "Random Forest": [10.8, 6.4, 6.0, 5.7, 5.2, 4.7, 4.4, 4.2, 3.9],
}
mape_B = {
    "Decision Tree": [635, 330, 340, 68, 45, 30, 25, 24, 22],
    "Extra Trees":   [215, 150, 152, 128, 82, 48, 33, 27, 22],
    "Random Forest": [480, 272, 193, 130, 45, 30, 26, 24, 22],
}

# ---- sanity check: every series matches its x-axis length ----
for name, d, n in [("A", [r2_A, rmse_A, mape_A], len(min_records)),
                   ("B", [r2_B, rmse_B, mape_B], len(ratio_labels))]:
    for metric in d:
        for m in MODELS:
            assert len(metric[m]) == n, f"Panel {name}: {m} has wrong length"
print("Data OK — all series match their x-axis length.")
````

## 2. Plotting helper

````python
def plot_metric(ax, x, data, ylabel, xlabel, logx=False, xticklabels=None):
    """Draw the three models on one axis with consistent colours/markers."""
    for m in MODELS:
        ax.plot(x, data[m], marker=MARKERS[m], color=COLORS[m], label=m,
                markeredgecolor="white", markeredgewidth=0.5,
                clip_on=False, zorder=3)
    if logx:
        ax.set_xscale("log")
        ax.set_xticks(x)
        ax.xaxis.set_major_formatter(FuncFormatter(lambda v, _: f"{int(round(v))}"))
        ax.xaxis.set_minor_formatter(NullFormatter())
        ax.minorticks_off()
    if xticklabels is not None:
        ax.set_xticks(x)
        ax.set_xticklabels(xticklabels, rotation=45, ha="right")
    ax.set_ylabel(ylabel)
    ax.set_xlabel(xlabel)
    ax.tick_params(direction="out", length=3)
    for s in ("top", "right"):
        ax.spines[s].set_visible(False)
````

## 3. Build the composite figure and export

````python
fig = plt.figure(figsize=(7.48, 6.3), layout="constrained")  # 190 mm wide
subfigs = fig.subfigures(3, 1, height_ratios=[1.0, 1.18, 0.12])

# ---------------- Panel A : minimum records per well ----------------
axA = subfigs[0].subplots(1, 3)
plot_metric(axA[0], min_records, r2_A,   r"Test $R^{2}$", "Min. records per well", logx=True)
plot_metric(axA[1], min_records, rmse_A, "Test RMSE (m)", "Min. records per well", logx=True)
plot_metric(axA[2], min_records, mape_A, "Test MAPE (%)", "Min. records per well", logx=True)
axA[0].set_ylim(0.3, 0.95)
subfigs[0].suptitle("(A)  Sensitivity to minimum records per well",
                    x=0.012, ha="left", fontsize=9, fontweight="bold")

# ---------------- Panel B : train / test ratio ----------------
axB = subfigs[1].subplots(1, 3)
plot_metric(axB[0], ratio_x, r2_B,   r"Test $R^{2}$", "Train / test ratio", xticklabels=ratio_labels)
plot_metric(axB[1], ratio_x, rmse_B, "Test RMSE (m)", "Train / test ratio", xticklabels=ratio_labels)
plot_metric(axB[2], ratio_x, mape_B, "Test MAPE (%)", "Train / test ratio", xticklabels=ratio_labels)
axB[0].set_ylim(-0.7, 1.0)
axB[0].axhline(0, color="0.6", lw=0.6, ls=(0, (3, 3)), zorder=1)  # R^2 = 0 reference
subfigs[1].suptitle("(B)  Sensitivity to train\u2013test ratio",
                    x=0.012, ha="left", fontsize=9, fontweight="bold")

# ---------------- shared legend (own band, no overlap) ----------------
handles = [Line2D([0], [0], color=COLORS[m], marker=MARKERS[m],
                  markeredgecolor="white", markeredgewidth=0.5, label=m)
           for m in MODELS]
subfigs[2].legend(handles=handles, loc="center", ncol=3, frameon=False,
                  handlelength=1.8, columnspacing=2.0)

# ---------------- export ----------------
stem = "Figure6_sensitivity_temporal_forecasting"
fig.savefig(OUTDIR / f"{stem}.pdf")                       # vector — submit this
fig.savefig(OUTDIR / f"{stem}.png", dpi=600)              # 600-dpi raster preview
fig.savefig(OUTDIR / f"{stem}.tif", dpi=600,
            pil_kwargs={"compression": "tiff_lzw"})       # 600-dpi LZW TIFF
print("Saved:",
      *[str(OUTDIR / f"{stem}{e}") for e in (".pdf", ".png", ".tif")], sep="\n  ")
plt.show()
````
